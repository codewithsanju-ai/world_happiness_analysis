```python
top10 = df.nlargest(10, "Happiness Score")

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top10,
    x="Happiness Score",
    y="Country",
    color="teal"
)

plt.title=("Top 10 Happiest Countries")
plt.xlabel=("Happiness Score")
plt.ylabel=("Country")

plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Happiness Score",
    bins=15,
    kde=True,
    color="teal"
)

plt.title=("Happiness Score Distribution")
plt.xlabel=("Happiness Score")
plt.ylabel=("number of Country")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

```python


# Region-wise average
result = df.groupby("Region")[
    ["Economy (GDP per Capita)", "Family",
     "Health (Life Expectancy)", "Freedom"]
].mean().reset_index()

print(result.round(2))

# Graph
result_melted = result.melt(
    id_vars="Region",
    var_name="Factor",
    value_name="Average Score"
)

plt.figure(figsize=(14, 7))

sns.barplot(
    data=result_melted,
    x="Region",
    y="Average Score",
    hue="Factor",
    palette="viridis"
)

ax = plt.gca()
ax.set_xlabel("Region")
ax.set_ylabel("Average Score")
ax.set_title("Average Regional Scores for Key Factors")
ax.tick_params(axis="x", labelrotation=45)
ax.legend(title="Factors")

plt.tight_layout()
plt.show()
```
